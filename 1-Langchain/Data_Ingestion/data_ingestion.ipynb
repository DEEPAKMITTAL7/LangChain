```python
import arxiv, os, requests

_client = arxiv.Client()

def _results(self):
    return _client.results(self)

def _download_pdf(self, dirpath="./", filename=""):
    path = os.path.join(dirpath, filename or f"{self.get_short_id()}.pdf")
    r = requests.get(self.pdf_url.replace("http://", "https://"))
    r.raise_for_status()
    with open(path, "wb") as f:
        f.write(r.content)
    return path

arxiv.Search.results = _results
arxiv.Result.download_pdf = _download_pdf
```
